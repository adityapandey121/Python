```python
## 1) The input() function always returns the user's input as a string (str) by default.
```

```python
age = input("Enter your age: ")
print(age)
print(type(age))
```

```python
##Why does Python do this?
#When you type on the keyboard, the computer receives characters (text), not numbers.
#Python has no way to know whether you intended those characters to represent:
#a number (123)
#a name ("123")
#a password ("123")
#a phone number ("1234567890")
#a ZIP code ("560001")

#Because of this ambiguity, Python safely returns everything as a string and lets you decide how to interpret it.
```

```python
##input() always returns a string because keyboard input is received as text (a sequence of characters). Python cannot reliably infer whether the user intended the input to be an integer, float, or plain text, so it returns a string by default. The programmer can then explicitly convert it using functions like int() or float() as needed.
```

```python
#Concentric Square
```

```python
n = 5
size = 2 * n - 1
for i in range(size):

    for j in range(size):
        top = i
        left = j
        bottom = size - 1 - i
        right = size - 1 - j
        minimum = min(top, left, bottom, right)
        print(n - minimum, end=" ")
    print()
```

```python
# Hour Glass
```

```python
n = 5

# Upper Half
for i in range(n, 0, -1):

    # Print spaces
    for j in range(n - i):
        print(" ", end="")

    # Print stars
    for j in range(2 * i - 1):
        print("*", end="")

    print()

# Lower Half
for i in range(2, n + 1):

    # Print spaces
    for j in range(n - i):
        print(" ", end="")

    # Print stars
    for j in range(2 * i - 1):
        print("*", end="")

    print()
```

```python
# Christmas Tree
```

```python
n = 5

# Top pyramid
for i in range(1, n + 1):

    # Spaces
    for j in range(n - i):
        print(" ", end="")

    # Stars
    for j in range(2 * i - 1):
        print("*", end="")

    print()


# Bottom pyramid
for i in range(3, n + 1):

    # Spaces
    for j in range(n - i):
        print(" ", end="")

    # Stars
    for j in range(2 * i - 1):
        print("*", end="")

    print()


# Trunk
for i in range(2):

    for j in range(n - 1):
        print(" ", end="")

    print("*")
```

```python
# Heart Pattern
```

```python
n = 6

# Upper Part
for i in range(n // 2, n + 1, 2):

    # Left Spaces
    for j in range(1, n - i + 1):
        print(" ", end="")

    # Left Stars
    for j in range(1, i + 1):
        print("*", end="")

    # Middle Spaces
    for j in range(1, n - i + 1):
        print(" ", end="")

    # Right Stars
    for j in range(1, i + 1):
        print("*", end="")

    print()


# Lower Part
for i in range(n, 0, -1):

    # Left Spaces
    for j in range(n - i):
        print(" ", end="")

    # Stars
    for j in range(2 * i - 1):
        print("*", end="")

    print()
```

```python
 n = 5

for i in range(1, n + 1):

    # Print spaces
    for j in range(n - i):
        print(" ", end="")

    # Increasing numbers
    for j in range(1, i + 1):
        print(j, end="")

    # Decreasing numbers
    for j in range(i - 1, 0, -1):
        print(j, end="")

    print()
```
